```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser, PydanticToolsParser
```

```python
prompt_template = PromptTemplate.from_template(
    "system: You are a helpful assistant that answers questions.\n"
    "user: {question}\n"
)

parser = StrOutputParser()

prompt = prompt_template.format(question= "What is the capital of France?")

prompt, parser
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert tech analyst. Provide a concise, 3-bullet summary of the topic provided."),
    ("human", "Topic: {topic}")
])

prompt
```

```python
chain = prompt | llm | parser
chain
```

```python
response = chain.invoke({"topic": "Artificial Intelligence"})
response
```

```python
from pydantic import BaseModel, Field 
import pprint
```

```python
class TechBrief (BaseModel):
    topic: str = Field(description="The technology topic to summarize.")
    summary: str = Field(description="A concise, 3-bullet summary of the topic.")
    key_insights: list[str] = Field(description="A list of key insights about the topic. Atlest 3 insights should be provided.")
    translation: str = Field(description="A translation of the summary in Spanish.")
    tech_topic: bool = Field(description="A boolean indicating whether the topic is a technology or not.")


structured_llm = llm.with_structured_output(TechBrief)
structured_llm
```

```python
prompt
```

```python
structured_chain = prompt | structured_llm
structured_chain
```

```python
response = structured_chain.invoke({"topic": "crops"})
response
```

```python
print("topic:", response.topic)
print("summary:", response.summary)
print("key_insights:", response.key_insights)
print("translation:", response.translation)
print("tech_topic:", response.tech_topic)
```
