```python
from pydantic import BaseModel, Field 
import pprint
```

```python
class TechBrief (BaseModel):
    topic: str = Field(description="The technology topic to summarize.")
    summary: str = Field(description="A concise, 3-bullet summary of the topic.")
    key_insights: list[str] = Field(description="A list of key insights about the topic. Atlest 3 insights should be provided.")
    translation: str = Field(description="A translation of the summary in Spanish.")
    tech_topic: bool = Field(description="A boolean indicating whether the topic is a technology or not.")


structured_llm = llm.with_structured_output(TechBrief)
structured_llm
```

```python
prompt
```

```python
structured_chain = prompt | structured_llm
structured_chain
```

```python
response = structured_chain.invoke({"topic": "crops"})
response
```

```python
print("topic:", response.topic)
print("summary:", response.summary)
print("key_insights:", response.key_insights)
print("translation:", response.translation)
print("tech_topic:", response.tech_topic)
```

```python
import math
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_core.tools import tool
```

```python
search_tool = DuckDuckGoSearchRun()
search_tool.invoke({"query": "What is the closest star to Earth?"})
```

```python
@tool
def Search_internet(query: str) -> str:
    """Search the internet for the given query and return a concise summary of the results."""
    search_tool = DuckDuckGoSearchRun()
    results = search_tool.invoke({"query": query})
    # For simplicity, we'll just return the titles of the top 3 results
    # summary = "\n".join([result["title"] for result in results[:3]])
    return results
```

```python
serch = Search_internet
search_result = serch.invoke("What is the closest star to Earth?")
```

```python
search_result
```

```python
@tool
def calculate_compound_interest(principal: float, rate: float, years: int) -> float:
    """
    Calculates the final amount after compound interest.
    Use this when you need to calculate investment returns.
    Args:
        principal: The initial amount of money
        rate: The annual interest rate as a decimal (e.g., 0.05 for 5%)
        years: The number of years the money is invested
    """
    return principal * math.pow((1 + rate), years)
```

```python
tools = [ Search_internet, calculate_compound_interest]
```

```python
llm_with_tools = llm.bind_tools(tools)
```

```python
msg = llm_with_tools.invoke([HumanMessage(content="If I invest $1000 at 5% for 10 years, what do I get?")])
msg
```
